# Google Sheets EDA

This notebook connects to a Google Sheet and performs basic Exploratory Data Analysis (EDA).

## Prerequisites
1.  **Service Account JSON Key**: obtain a service account JSON key from the Google Cloud Console.
2.  **Share the Sheet**: Share your target Google Sheet with the `client_email` address found in your JSON key file.

```python
import pandas as pd
import gspread
from google.oauth2.service_account import Credentials
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
```

## Authentication & Connection

Replace `SERVICE_ACCOUNT_FILE` with the path to your JSON key file and `SHEET_NAME` with the name of your Google Sheet.

```python
# Configuration
SERVICE_ACCOUNT_FILE = 'credentials.json'  # Path to your service account key file
SHEET_NAME = 'My Sheet Name'               # Name of the Google Sheet
WORKSHEET_INDEX = 0                        # Index of the worksheet (0 is the first one)

# Scopes required
SCOPES = [
    'https://www.googleapis.com/auth/spreadsheets',
    'https://www.googleapis.com/auth/drive'
]

try:
    creds = Credentials.from_service_account_file(SERVICE_ACCOUNT_FILE, scopes=SCOPES)
    client = gspread.authorize(creds)
    
    # Open the spreadsheet
    sheet = client.open(SHEET_NAME)
    
    # Select worksheet
    worksheet = sheet.get_worksheet(WORKSHEET_INDEX)
    
    print(f"Successfully connected to: {sheet.title}")
except FileNotFoundError:
    print(f"Error: The file {SERVICE_ACCOUNT_FILE} was not found. Please ensure you have placed your service account JSON file in the project directory.")
except Exception as e:
    print(f"An error occurred: {e}")
```

## Load Data

```python
# Get all records and convert to DataFrame
data = worksheet.get_all_records()
df = pd.DataFrame(data)

# Display first few rows
df.head()
```
